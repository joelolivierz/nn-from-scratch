```python
import numpy as np
```

```python
class Network:
    
    def __init__(self, dimensions: tuple):
        self.dimensions = dimensions
        self.amount_of_layers = len(dimensions)
        dimensions = tuple(enumerate(dimensions))
        for dimension in dimensions:
            layer_numb, layer_dim = dimension
            if layer_numb == 0:
                setattr(self, f"input_layer", np.full(layer_dim, 0.5))
            if layer_numb == len(dimensions)-1:
                setattr(self, f"output_layer", np.full(layer_dim, 0.5))
            else:
                setattr(self, f"layer_{layer_numb}", np.full(layer_dim, 0.5))
    
    def __str__(self):
        return "Neural Network: \n" + f"\tHidden Layers: {self.amount_of_layers-2}\n" + "\tDimensions " + " | ".join(str(dim) for dim in self.dimensions)

    def __repr__(self):
        return "Network"
```
